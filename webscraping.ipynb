```python
import ipywidgets as widgets
from IPython.display import display, HTML
import threading
import time
import feedparser

# Styling
style = """
    <style>
        .widget-button { background-color: #007BFF; color: white; font-size: 16px; }
        .widget-text { width: 400px; font-size: 14px; }
        .progress-bar { height: 10px; background-color: #007BFF; transition: width 0.5s; }
        .progress-container { width: 100%; background-color: #e0e0e0; height: 10px; }
        .output-box { border: 1px solid #ddd; padding: 10px; margin-top: 10px; background: #f9f9f9; }
    </style>
"""
display(HTML(style))

# Widgets
url_input = widgets.Text(placeholder="Enter RSS Link to Scrape", layout=widgets.Layout(width='400px'))
scrape_button = widgets.Button(description="Start Scraping", button_style='primary')
progress = widgets.IntProgress(value=0, min=0, max=100, step=20, bar_style='info')
output_area = widgets.Output()


def scrape_website(url):

    with output_area:
        output_area.clear_output()
        display(HTML(f'<p style="color:black; font-size:16=8px;font_family:Fort">Scraping started for: {url}</p>'))

    feed = feedparser.parse(url)

    feed = feedparser.parse(url)

    data = []

    for entry in feed.entries:
        print(entry)
        data.append(entry)
    df = pd.DataFrame(data)

    # with output_area:
    #     display(HTML(f'<div class="output-box"><strong>{entry.title}</strong><br><a href="{entry.link}" target="_blank">{entry.link}</a></div>'))
    # with output_area:
    #         display(HTML(f"<h3>Scraped Data:</h3>"))
    #         display(df)

    # with output_area:
    #     display(HTML(f'<p style="color:green; font-size:16px;">Scraping complete for: {url}</p>'))
    return df


def start_scraping(_):
    progress.value = 0
    output_area.clear_output()

    # Start the scraping in a separate thread
    def run_scraping():
        df = scrape_website(url_input.value)
        with output_area:
            display(HTML(f"<h3>Scraped Data:</h3>"))
            display(df)

    thread = threading.Thread(target=run_scraping())
    thread.start()

scrape_button.on_click(start_scraping)

# Display UI
display(widgets.VBox([url_input, scrape_button, progress, output_area]))
```

```python
import streamlit as st
import altair as alt
import pandas as pd
from vega_datasets import data


@st.cache_data
def get_data():
    source = data.stocks()
    source = source[source.date.gt("2004-01-01")]
    return source


stock_data = get_data()

hover = alt.selection_single(
    fields=["date"],
    nearest=True,
    on="mouseover",
    empty="none",
)

lines = (
    alt.Chart(stock_data, title="Evolution of stock prices")
    .mark_line()
    .encode(
        x="date",
        y="price",
        color="symbol",
    )
)

points = lines.transform_filter(hover).mark_circle(size=65)

tooltips = (
    alt.Chart(stock_data)
    .mark_rule()
    .encode(
        x="yearmonthdate(date)",
        y="price",
        opacity=alt.condition(hover, alt.value(0.3), alt.value(0)),
        tooltip=[
            alt.Tooltip("date", title="Date"),
            alt.Tooltip("price", title="Price (USD)"),
        ],
    )
    .add_selection(hover)
)

data_layer = lines + points + tooltips

ANNOTATIONS = [
    ("Sep 01, 2007", 450, "🙂", "Something's going well for GOOG & AAPL."),
    ("Nov 01, 2008", 220, "🙂", "The market is recovering."),
    ("Dec 01, 2007", 750, "😱", "Something's going wrong for GOOG & AAPL."),
    ("Dec 01, 2009", 680, "😱", "A hiccup for GOOG."),
]
annotations_df = pd.DataFrame(
    ANNOTATIONS, columns=["date", "price", "marker", "description"]
)
annotations_df.date = pd.to_datetime(annotations_df.date)

annotation_layer = (
    alt.Chart(annotations_df)
    .mark_text(size=20, dx=-10, dy=0, align="left")
    .encode(x="date:T", y=alt.Y("price:Q"), text="marker", tooltip="description")
)

combined_chart = data_layer + annotation_layer
st.altair_chart(combined_chart, use_container_width=True)
```
